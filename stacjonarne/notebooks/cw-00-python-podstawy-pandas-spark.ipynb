# 🧩 Część A: Python — listy, słowniki i funkcje

## 📖 Przykład — przegląd operacji na listach i słownikach

Poniżej kompletny przykład, z którego będziesz korzystać przy rozwiązywaniu ćwiczeń.

````python
# 📖 Przykład: oceny studentów grupy "Python 101"

studenci = ["Ada", "Alan", "Grace", "Linus", "Grace"]
oceny = [5.0, 3.5, 4.5, 4.0, 4.5]

# Zip + dict comprehension -> słownik imię -> ocena
oceny_slownik = {imie: ocena for imie, ocena in zip(studenci, oceny)}
print(oceny_slownik)

# List comprehension -> tylko oceny >= 4.5
najlepsi = [imie for imie, ocena in oceny_slownik.items() if ocena >= 4.5]
print("Najlepsi:", najlepsi)

# Counter -> ile razy każde imię się pojawia na liście
print(Counter(studenci))
````
